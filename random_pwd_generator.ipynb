```python
import math
import secrets
import string

MIN_LENGTH = 8      
MAX_LENGTH = 128    
NIST_LENGTH = 15    

# ---------- PHASE 1: INPUT (collect and validate) ----------
def get_length():
    while True:
        text = input(f"\nEnter password length ({MIN_LENGTH} - {MAX_LENGTH}): ").strip()
        try:
            length = int(text)
        except ValueError:
            print("Invalid input. Please enter a whole number like 16.")
            continue
        if length < MIN_LENGTH or length > MAX_LENGTH:
            print(f"Length must be between {MIN_LENGTH} and {MAX_LENGTH}.")
            continue
        return length
        
def ask_yes_no(question):
    while True:
        answer = input(question + " (y/n): ").strip().lower()
        if answer in ("y", "yes"):
            return True
        if answer in ("n", "no"):
            return False
        print("Please type y or n.")
        
# ---------- PHASE 2: PROCESS (the transformation engine) ----------
def build_pool(use_symbols):
    pool = string.ascii_letters + string.digits     
    if use_symbols:
        pool += string.punctuation                
    return pool
    
def meets_rules(password, use_symbols):
    has_letter = any(c in string.ascii_letters for c in password)
    has_digit = any(c in string.digits for c in password)
    has_symbol = any(c in string.punctuation for c in password)
    return has_letter and has_digit and (has_symbol or not use_symbols)

def generate_password(length, use_symbols):
    pool = build_pool(use_symbols)
    while True:
        characters = [secrets.choice(pool) for _ in range(length)]  
        password = "".join(characters)                              
        if meets_rules(password, use_symbols):
            return password

# ---------- PHASE 3: OUTPUT (prove the security) ----------
def calculate_entropy(length, pool_size):
    return length * math.log2(pool_size)
    
def rate_strength(entropy):
    if entropy < 40:
        return "Weak"
    if entropy < 60:
        return "Fair"
    if entropy < 80:
        return "Strong"
    return "Very strong"

def show_result(password, length, use_symbols):
    pool_size = len(build_pool(use_symbols))
    entropy = calculate_entropy(length, pool_size)
    print("\n=============================")
    print(f"Password  : {password}")
    print(f"Length    : {length}")
    print(f"Pool size : {pool_size} characters")
    print(f"Entropy   : {entropy: .1f} bits ({rate_strength(entropy)})")
    print("=============================")
    if length < NIST_LENGTH:
        print(f"Tip: NIST 2024 recommends {NIST_LENGTH} + characters for high-security use.")
        
# ---------- MAIN ----------
def main():
    print("===== RANDOM PASSWORD GENERATOR =====")
    while True:
        length = get_length()
        use_symbols = ask_yes_no("Include special characters (@, #, $ ...)?")
        password = generate_password(length, use_symbols)
        show_result(password, length, use_symbols)
        if not ask_yes_no("\nGenerate another password?"):
            print("Goodbye! Keep your passwords safe.")
            break
            
if __name__ == "__main__":
    main()
```
